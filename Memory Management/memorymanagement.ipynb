 ## Best Practices 
 
 use local Variable 

 Avoid circular refrences = b=1 a= c like that 

 use gnerator- becs produces one item at a time 

 explctly  Delete Objects 

 profile memory Usage 

```python
import gc

class MyObeject:
    def __init__(self,name):
        self.name=name
        print(f'object {self.name} created')
    
    def __del__(self):
        print(f'object {self.name} deleted')

## circular  refrence 
obj1=MyObeject('obj1')
obj2=MyObeject('obj2')
obj1.ref=obj2
obj2.ref=obj1

del obj1
del obj2 
print(gc.garbage)
##manually trigger the collection becz  both  are circular so not automatically collected 
gc.collect() 

##printing collected objects 
print(gc.garbage)
```

```python
## genrators for memory efficiency 
## genrator allow you to produyce items one at a time , using memory efficiency by keeping only one itterator 

def generate_number(n):
    for i in range(n):
        yield i 

## using the genrator 

for num in generate_number(1000):
    print(num)
    if num>10 : 
        break
    
```
