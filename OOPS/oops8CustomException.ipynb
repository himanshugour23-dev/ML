Custom Exception

```python
class Error(Exception):
    """Base class for other exceptions"""
    pass


class DateOfBirthException(Exception):
    """Raised when the input date of birth is invalid"""
    pass


year = int(input("Enter your year of birth: "))
age = 2024 - year

try:
    if 20 <= age <= 30:
        print("You are eligible for the job")
    else:
        raise DateOfBirthException

except DateOfBirthException:
    print("You are not eligible for the job")
```
