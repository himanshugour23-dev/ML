POLYMORPHISM

Allow object of different classes to be treated as object of common superclass 
piolymorphism is achieved through the overriding and interfaces

Method overriding

This allows the specific implimentation of a method alerady defined in its parent calss 

```python
##Base Class 
class Animal:
    def speak(self):
        return "Animal speaks"
##Derived Class
class Dog(Animal):
    def speak(self):
        return "Dog barks"
## same method in child and parent class but different implementation is called method overriding.
class Cat(Animal):
    def speak(self):
        return "Cat meows"
##function that demonstrates polymorphism
def animal_speak(animal):
    print(animal.speak())
## animal hear above fn is object of class and it can be of any class that is derived from Animal class.
dog = Dog() 
cat = Cat()
print(dog.speak())  # Output: Dog barks
print(cat.speak())  # Output: Cat meows
print(animal_speak(dog))  # Output: Dog barks
```

```python
## polymorphism with functions and methods
##base class
class Shape:
    def area(self):
        return "The area of the figure "
##derived class
class Rectangle(Shape):
    def __init__(self,width,height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height
##derived class 2 
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

##function that demonstrates polymorphism
def calculate_area(shape):
    print(shape.area())

rectangle = Rectangle(5, 10)
circle = Circle(7)

calculate_area(rectangle)
calculate_area(circle)
```

Polymorphism with Abstract Base Class

ABCs are common method for a group related objects . they can enforce that derived classes ompliment particular methods , promoting consistency across different implimentations. 

```python
from abc import ABC, abstractmethod 

##define abstract class 
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

##derived class 1
class Car(Vehicle):
    def start_engine(self):
        return "Car engine started"
##derived class 2
class Motorcycle(Vehicle):
    def start_engine(self):
        return "Motorcycle engine started"

## create objects of car and motorcycle
car = Car()
motorcycle = Motorcycle()
## function that demonstrates polymorphism
def start_sound(vehicle):
    print(vehicle.start_engine())
start_sound(car)
start_sound(motorcycle)
```
