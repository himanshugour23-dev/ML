Abstraction 

Abstraction is the concept of hiding the complex implimentation details showing only necessary feature of bobject . 

```python
from abc import ABC, abstractmethod
## when we use abstract method we make suyre fn is empty and we have to override it in derived class otherwise we will get error
##abstract base class 
class Vehicle(ABC):
    def drive(self):
        print("The vehical is used for driving") 
    
    @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        print("The car engine is started")

def operate_vehicle(vehicle):
    vehicle.start_engine()
    vehicle.drive()
my_car = Car()
operate_vehicle(my_car)
```
