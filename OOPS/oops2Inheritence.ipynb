Inheritence  
allow to inherit the atteribute from another class 

```python
##Inheritence 
##Parent Class 
class Car:
    def __init__(self,windows,doors,engineType):
        self.windows = windows
        self.doors = doors
        self.engineType = engineType
    def drive(self):
        print(f"Car is drive the {self.engineType} engine Car")
    
```

```python
## Inheriting Car Class
##super() is used to call the parent class constructor and its methods in the child class.
class Tesla(Car):
    def __init__(self,windows,doors,engineType,autopilot):
        super().__init__(windows,doors,engineType)
        self.autopilot = autopilot
    def autopilot_feature(self):
        print(f"Tesla has autopilot feature: {self.autopilot}")

musk = Tesla(4,4,"Electric Trash",True)
musk.autopilot_feature()
musk.drive()
```
