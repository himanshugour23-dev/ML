```python
## file handeling and Exceptipon Handeling
try : 
    file = open("test.txt","r")
    content = file.read() 
    print(content)
except FileNotFoundError as ex:    ## file not  there error
    print(ex)
finally:
    if 'file' in locals() and not file.closed() :
        file.close()
        print("file is closed")
```
