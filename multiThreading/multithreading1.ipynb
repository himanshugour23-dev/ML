When to use Multithreading

Its when Task spend more time waiting for I/O operations

And have concurrent applications at Singel time 

```python
import threading
import time 
```

```python
def print_number():
    for i in range(5):
        print(f"Number{i}") ; 

def print_letters():
    for l in "abcde":
        print(f"Letter {l}")

t = time.time() ; 
print_number()
print_letters() 
finshed_time = time.time()  - t ; 

finshed_time
```

```python
def print_number():
    for i in range(5):
        time.sleep(2)
        print(f"Number{i}") ; 

def print_letters():
    for l in "abcde":
        time.sleep(2)
        print(f"Letter {l}")

#t = time.time() ; 
#print_number()
#print_letters() 
#finshed_time = time.time()  - t ; 

#finshed_time

## so when printing numbert is on sleep i wanted to make sure that other function excute so its tie to use multithreading 
```

```python
## multithreading it so it can run simultaneously 

t1 = threading.Thread(target=print_number)
t2 = threading.Thread(target=print_letters)
t1.start()
t2.start()

## wait for the threads to finish so need to use finsh 
t1.join() 
t2.join() 
finshed_time = time.time()  - t ; 
print(finshed_time) ; finshed_time

## During sleep if one Thread is on Sleep than and than only 
```
