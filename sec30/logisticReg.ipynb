```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
##if we are creating a dataSet From This method wE do not need a standerd scaler because the data is already scaled and normalized.
from sklearn.datasets import make_classification
X,y = make_classification(n_samples=1000, n_features=2, n_classes=2, n_redundant=0, random_state=42)
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.30,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
## C is inversely Propotional to lembda (lembda of Rigid lasso)
##ClassWeight is used to handle the imbalanced data set. It will give more weight to the minority class and less weight to the majority class.
logistic = LogisticRegression()
```

```python
logistic.fit(X_train,y_train)
```

```python
y_pred= logistic.predict(X_test)
y_pred
```

```python
logistic.predict_proba(X_test)
```

```python
from sklearn.metrics import confusion_matrix, classification_report,accuracy_score
```

```python
score = accuracy_score(y_test,y_pred)
print(score)
cm = confusion_matrix(y_test,y_pred)
print(cm)
cr = classification_report(y_test,y_pred)
print(cr)
```

```python
## We Will Learn hyperParameter tuning in the next section. We will use GridSearchCV to find the best hyperparameters for our model.
## parameter such as C, class_weight, solver, max_iter, etc. We will also use cross-validation to find the best hyperparameters for our model.
```

HyperParameter Tuning 

```python
model  = LogisticRegression() 
penalty = ['l1','l2','elasticnet','none']
c_values = [0.001,0.01,0.1,1,10,100]
solver = ['newton-cg','lbfgs','liblinear','sag','saga']
```

```python
params = [
    {
        'penalty': ['l1'],
        'C': c_values,
        'solver': ['liblinear', 'saga']
    },
    {
        'penalty': ['l2'],
        'C': c_values,
        'solver': ['lbfgs', 'liblinear', 'sag', 'saga']
    }
]
```

```python
print(type(params))
```

```python
##GridSearchCV is a method to find the best hyperparameters for our model. It will try all the combinations of hyperparameters and return the best combination of hyperparameters.
##GridSearchCV used tofind 
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import StratifiedKFold
cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)
grid = GridSearchCV(
    model,
    params,
    cv=cv,
    n_jobs=-1
)

grid.fit(X_train, y_train)

print(grid.best_params_)
print(grid.best_score_)
```

```python
grid
```

```python
## Now it will try to find the Best One That exist 
grid.fit(X_train,y_train)
```

```python
y=grid.predict(X_test)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
score = accuracy_score(y_test,y)
print(score)
cm = confusion_matrix(y_test,y)
print(cm)
cr = classification_report(y_test,y)
print(cr)
```
