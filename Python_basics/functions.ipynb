FUNCTIONS IN  PYTHON 

```python
def add (a,b):
    return a+b ; 

print(add(2,3)) 
```

```python
##default values 
def greet(name="Noname"):
    return f"Hello, {name}!"
print(greet())
print(greet("Alice"))
```

```python
##variable Length arguments
##Positional and keyWord 

def print_numbers(*args):
    for i in args : 
        print(i)
print_numbers(1,2,3,4,5)
```

```python
### Keywords arguments
## alll parameter are in form of key val pair 

def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_info(name="Alice", age=30, city="New York")
```
