```python
## Creation of Dictionary 
# # Way 1
empty_dict = {}
print(type(empty_dict))
```

```python
# Way 2
empty_dict = dict()
empty_dict
```

```python
## key value pair  
## similarly to map wps se key dalege to replace hojyenga 
my_dict = {"name": "John", "age": 30, "city": "New York"}
print(my_dict)
```

```python
#accessing Dictionary elements  just give key name 
print(my_dict["name"])

## using the get method  give the key name there 
print(my_dict.get("age"))

## giving default val if key is not present
print(my_dict.get("country", "Not given baby "))
```

```python
##Modifying Dictionary Elements add update or delete
# Adding a new key-value pair
my_dict["country"] = "USA"
print(my_dict)
# Updating an existing key-value pair
my_dict["age"] = 31
print(my_dict)
# Deleting a key-value pair
del my_dict["city"]
print(my_dict)
```

```python
## dictionary Methods 
# Get all keys
keys  = my_dict.keys()
print(keys)

# Get all values
values = my_dict.values()
print(values)

# Get all key-value pairs
items = my_dict.items()
print(items)
```

```python
## shalllow copy
# Create a shallow copy of the dictionary
shallow_copy = my_dict.copy()
print(shallow_copy)
## others way to copy
shallow_copy2 = my_dict
print(shallow_copy2)
```

```python
## deepCopy
import copy
# Create a deep copy of the dictionary
deep_copy = copy.deepcopy(my_dict)
print(deep_copy)
```

```python
##iterating over dictionary
for key in my_dict.keys():
    print(key)
for key, value in my_dict.items():
    print(f"{key}: {value}")
for value in my_dict.values():
    print(value)
for i in my_dict.keys():
    print(i)
```

```python
##nested dictionary


nested_dict = {
    "name": "John",
    "age": 30,
    "address": {
        "street": "123 Main St",
        "city": "New York",
        "state": "NY"
    }
}

print(nested_dict["address"]["city"])
```

```python
##itterating over nested dictionary
for key, value in nested_dict.items():
    if isinstance(value, dict):
        print(f"{key}:")
        for sub_key, sub_value in value.items():
            print(f"  {sub_key}: {sub_value}")
    else:
        print(f"{key}: {value}")
```

```python
##dictionary comprehension
squared_dic = {x:x**2 for x in range(5)}
print(squared_dic)
```

```python
##conditional dictionary comprehension
even_squared_dic = {x:x**2 for x in range(10) if x % 2 == 0}
print(even_squared_dic)
```

```python
number = [1, 2, 3, 4, 5,3,4,32,4,5,2,3,1,1]
freq = {} 
for i in number : 
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1
print(freq)
```
