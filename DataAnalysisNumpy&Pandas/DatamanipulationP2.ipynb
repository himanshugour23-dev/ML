```python
import pandas as pd 
df = pd.read_csv('data.csv')
df.head(5) ## first 5 rows of the data frame
##df.tail(5) ## last 5 rows of the data frame
```

```python
df.describe() ## summary statistics of the data frame
```

```python
df.info() ## information about the data frame
```

```python
df.isnull()    
df.isnull().sum() ## count of missing values in each column
```

```python
##handeling missing values
df.dropna(inplace=True) ## drop rows with missing values
df.fillna(0, inplace=True) ## fill missing values with 0
df.isnull().any(axis=1) ## check if there are any missing values in the data frame on column level
```

```python
##filling missing values with mean of the column 
df['SalaryFillna'] = df['Salary'].fillna(df['Salary'].mean()) ## create a new column SalaryFillna with filled missing values in Salary column with mean of the Salary column
##df['Salary'].fillna(df['Salary'].mean(), inplace=True) ## fill missing values in Salary column with mean of the Salary column
df
```

```python
df = df.rename(columns = {'SalaryFillna':'SalaryFilled'}) ## rename the column SalaryFillna to SalaryFilled
df
```

```python
##Change DataTypes 
df['Value_new'] = df['SalaryFilled'].astype(float) ## create a new column Value_new with changed data type of Value column to float
## astype not work  when there is NnN or any value not present so just use  fillna with mean or 0 to handle missing values before changing data type
df
```

```python
df['new_Salary'] = df['Salary'].apply(lambda x:x*10/100)
df
```

```python
## grouping of objects 
grouped = df.groupby('City')['Salary'].mean() ## group the data frame by City column
grouped
```

```python
## aggregate multiple fucntion 
agg = df.groupby('City')['Salary'].agg(['mean','sum','max','min'])
agg
```

```python
## merging and joining of data frames
df1 = pd.DataFrame({'Key':['A','B','C'],'Value1':[1,2,3]})
df2 = pd.DataFrame({'Key':['A','B','D'],'Value2':[5,6,7]})
```

```python
## Merge Dataframe on a common key  How joins are same as SQL joins
pd.merge(df1,df2,on='Key',how='inner')
```
