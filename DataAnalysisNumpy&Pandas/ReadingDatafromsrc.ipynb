Reading Data from srcses 

```python
import pandas as pd
from io import StringIO

Data = '{"employee_name": "James", "email": "james@gmail.com", "job_profile": [{"title":"Team Lead"}]}'
df = pd.read_json(StringIO(Data))
df
```

```python
df.to_json() ## convert the data frame to json format
df.to_json(orient='records') ## convert the data frame to json format with orient as records
```

```python
df.to_json(orient='index') ## convert the data frame to json format with orient as index
```

```python
df.to_csv('employee_data.csv',index=False) ## convert the data frame to csv format and save it to a file named employee_data.csv
```
