```python
import pandas as pd 

data = [1,2,3,4,5]
series = pd.Series(data)
print(series)

##series from dictionary
data = {'a' :1 , 'b' : 2 , 'c' : 3}
series_dic = pd.Series(data) 
print(series_dic)

data = [10,20,30] 
index = ['a','b','c']
pd.Series(data,index=index)
```

```python
##DataFrame
##ccreating form Dic 
import pandas as pd 
dict = {
    'Name' : ['himanshu' , 'john' , 'jack'],
    'Age' : [25,30,45],
    'city' : ['Banglore' , 'NewYork','London']
}
dict2 = [
    {'Name' : 'himanshu', 'Age' : 32 , 'City' : 'Banglore'  },
    {'Name' : 'himanshu', 'Age' : 32 , 'City' : 'Banglore'  },
    {'Name' : 'himanshu', 'Age' : 32 , 'City' : 'Banglore'  },
    {'Name' : 'himanshu', 'Age' : 32 , 'City' : 'Banglore'  },
]
dp = pd.DataFrame(dict)
##print(dp)

dp2 = pd.DataFrame(dict2)
##print(dp2)

##dp.iloc[0]      ## iloc =  index location 
##dp.iloc[0,2]  
##dp['Name']
print(dp.at[1,'Age'])     ##from first idx get age 
## Accessing specified element  at and iat
print(dp.iat[2,2])

## adding particular column 
dp['Salary'] = [200,300,400]
dp
## removing column  axis 0 - row check 
## axis1 col check   becz by defaault row check only at drop opn 
#dp.drop('Salary')  error as no salary col in rows 
dp.drop('Salary',axis=1,inplace=True)    ##inplace = True for saving otherwise bs dikhayega nnot save it 
dp.drop(1)  ## droped row idx 1 
```
