```python
from sklearn.datasets import fetch_california_housing
```

```python
import numpy as np
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
california = fetch_california_housing()
```

```python
converted = pd.DataFrame(california.data,columns=california.feature_names)
```

```python
converted
```

```python
converted['MedHouseVal'] = california.target
```

```python
converted
```

```python
converted.corr()
```

```python
## Independent and Dependent Features 
converted.head 
```

```python
X = converted.iloc[:,:-1] 
y  = converted.iloc[:,-1]
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.35, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler() 

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
##model Training 
from sklearn.linear_model import LinearRegression
linreg=LinearRegression()
linreg.fit(X_train,y_train)
```

```python
linreg.coef_
```

```python
##model Evaluation
y_pred = linreg.predict(X_test)

from sklearn.metrics import r2_score,adjusted_r2_score
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
print(mean_squared_error(y_test,y_pred))
print(mean_absolute_error(y_test,y_pred))
print(np.sqrt(mean_squared_error(y_test,y_pred)))
print(r2_score(y_test,y_pred))
```

```python
features = converted.drop('MedHouseVal', axis=1).columns

for feature in features:
    plt.figure()
    plt.scatter(converted[feature], converted['MedHouseVal'])
    plt.xlabel(feature)
    plt.ylabel('MedHouseVal')
    plt.show()
```
