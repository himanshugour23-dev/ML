## stamming 

```python
## Classification Problem
## Wheather the product has positive or negative reviews  
## Theory On my Copy 
## eating eat eaten -> eat is stem  other all of them are same stem 
## Ex : going go goes 
```

```python
words = ["eat", "eating", "eaten", "go", "going", "goes", "played", "playing", "plays", "play"]
```

Porter Stemming

```python
from nltk.stem import PorterStemmer
PorterStemmer = PorterStemmer()
for i in words:
    print(i+"------>"+PorterStemmer.stem(i))
```

```python
PorterStemmer.stem("congratulation")
```

```python
## stemming changing the meaning its cons  work for some word but not for all 
```

---> RegexStemmer Class   

 Take  single Expression and removes prefix or suffix that matches that expression

```python
from nltk.stem import RegexpStemmer
```

```python
regexp_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
regexp_stemmer.stem("congratulation")
```

Snowball Stema 

```python
from nltk.stem import SnowballStemmer
```

```python
snowball_stemmer = SnowballStemmer("english")
```

```python
for i in words:
    print(i+"------>"+snowball_stemmer.stem(i))
    
```

## Lemetisation 

Technique like stemming . o..p is called as lemma  after lematisation we get the correct gramatical form of that word  Kind oF dictionary  Helps us in it 

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
import nltk
```

```python
nltk.download('wordnet')
```

```python
'''
POS Noun-n
Verb-v
Adjective-a
Adverb-r
lemmatizer.lemmatize("better",pos="a")
'''

for i in words:
    print(i+"------>"+lemmatizer.lemmatize(i))
```
