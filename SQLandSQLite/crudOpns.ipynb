```python
import sqlite3
```

```python
##without any opns we are able to create it 
connection=sqlite3.connect('test.db')
```

```python
cursor=connection.cursor()
```

```python
##create table
## Create a Table

cursor.execute('''
CREATE TABLE IF NOT EXISTS employees(
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    age INTEGER,
    department TEXT
)
''')
##commiting the changes to the database
connection.commit()
```

```python
cursor.execute(''' select * from employees''')
```

```python
##insert the data into table 
cursor.execute(''' 
    Insert Into employees(name,age,department) values('John Doe',30,'HR')
''')
```

```python
## Query the data
cursor.execute(''' select * from employees''')
rows = cursor.fetchall()

## print 
for i in rows :
    print(i)
```

```python
## update the data in the table 
cursor.execute(
    '''
    UPDATE employees
    SET age = 31
    WHERE name = 'John Doe'
    '''
)

connection.commit()
```

```python
## Query the data
cursor.execute(''' select * from employees''')
rows = cursor.fetchall()

## print 
for i in rows :
    print(i)
```

```python
##Delete the data from the table
cursor.execute(
    '''    DELETE FROM employees
    WHERE name = 'John Doe' 
    ''')
connection.commit()
```
