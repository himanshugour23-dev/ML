File HAndeling Read and Write Files 
python Provides BuiltIn functions and methods to read and write to files both text and bonary 

```python
## Read a whole file 

with open('example.txt','r') as file : 
    content=file.read()
    print(content)
```

```python
##reading  file line by line 

with open('example.txt','r') as file : 
    for line in file : 
        print(line)
```

```python
#writing a file 
with open('example.txt','w') as file : 
    file.write('I am bhoomi \n' )
    file.write('I will kill himanshu\n' )
```

```python
## writing a file without overwriting 
with open('example.txt','a') as file :
    file.write('still i will love you bhoomi\n ')
```

```python
##writing a list of lines 
lines = [ 'first line and second line' , 'third line' ]
with open ('example.txt','a') as file :
    file.writelines(lines)
```

```python
##binary files-> we usually upddate bytes  .bin fules 
## wb is write byte mode and rb is read byte 
data =b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file :
    file.write(data)
```

```python
## reading binary files
with open('example.bin','rb') as file :
   content =  file.read()
   print(content)
```

IN w+ mopde you are able tio read and write both if file is nOt available it will create the File

```python
with open('example.txt', 'w+') as file:
    file.write('I am Himanshu\n')
    file.write('Please Kill me I not Need That much of pain anymore\n')

    # move cursor to start of file
    file.seek(0)

    content = file.read()
    print(content)
```
